```python
px.histogram(df,x='person_age',color='person_gender')

```

```python
# Insight:
# The age distribution shows the number of applicants across different age groups, with a comparison between male and female applicants.
```

```python
gender = df['person_gender'].value_counts()

plt.pie(
    gender.values,
    labels=gender.index,
    autopct='%1.1f%%'
)

plt.title('Gender Distribution')
plt.show()
```

```python
# Insight:
# The dataset contains both male and female applicants and male representing a larger proportion of the dataset.
```

```python
df['person_education'].unique()
```

```python
sns.countplot(x='person_education',data=df)
plt.title('Education Distribution')
plt.show()
```

```python
# Insight:
# The applicants have different education levels and Bachelor being more represented than others.
```

```python
intent=df['loan_intent'].value_counts().sort_values(ascending=False)
px.histogram(intent,x=intent.index,y=intent.values,text_auto='0f')
```

```python
# Insight:
# The distribution shows that applicants request loans for different purposes and education being more common than others.
```

```python
plt.pie(df['loan_status'].value_counts(),labels=df['loan_status'].unique(),autopct='%1.1f%%')
plt.title('Loan Status Distribution')
plt.show()
```

```python
# Insight:
# The dataset contains both approved and non-approved loan applications and approved representing a larger share of the observations.
```

```python
fig = px.scatter(df,x='person_income',y='loan_amnt',color='loan_status',trendline='ols',title='Income vs Loan Amount by Loan Status')
fig.show()
```

```python
# Insight:
# The visualization shows the relationship between personal income and loan amount, while loan status helps compare approved and non-approved applications.
```

```python
group= df.groupby(
    ['person_education', 'person_gender'],
    as_index=False
)['person_income'].mean()

fig = px.bar(group,x='person_education',y='person_income',color='person_gender', barmode='group',text_auto='.0f',title='Average Income by Education and Gender')
fig.show()
```

```python
# Insight:
# Average income varies across education levels and genders.
# The chart allows comparison of income patterns between these groups.
```

```python
group=df.groupby('person_home_ownership')['loan_amnt'].mean().reset_index()
fig=px.bar(group,x='person_home_ownership',y='loan_amnt',text_auto='.0f',title='Average Loan Amount by Home Ownership')
fig.show()
```

```python
# Insight:
# Average loan amounts differ across home ownership categories, showing variation in borrowing patterns between applicants.
```

```python
px.scatter(df, x='loan_amnt', y='loan_int_rate', color='loan_status').show()
```

```python
# Insight:
# The chart shows how interest rates vary across different loan amounts,  with loan status providing an additional comparison between applications.
```

```python
px.box(
    df,
    x='loan_status',
    y='loan_percent_income',
    title='Loan Percent Income vs Loan Status'
).show()
```

```python
# Insight:
# The credit score distributions differ between loan status groups,allowing us to compare the credit profiles of the applicants.
```

```python
px.scatter(df,x='person_age',y='person_emp_exp',color='person_gender',title='Age vs Employment Experience by Gender')
```

```python
# Insight:
# Employment experience generally increases with age,
# although there is variation among applicants.
```

```python
group = df.groupby(
    'previous_loan_defaults_on_file'
)['loan_status'].mean().reset_index()

group['loan_status'] = group['loan_status'] * 100
group
```

```python
group = df.groupby(
    ['person_age', 'person_gender'],
    as_index=False
)['loan_status'].mean()
fig = px.scatter(
    group,
    x='person_age',
    y='loan_status',
    color='person_gender',
    title='Loan Approval Rate by Age and Gender'
)

fig.show()
```

```python
# Insight:
# Loan approval rates vary across different age groups and genders,although the number of applicants may differ between groups.
```

```python
corr = df.select_dtypes(include='number').corr()
plt.figure(figsize=(12, 8))
sns.heatmap(corr, annot=True)

plt.show()
```

```python
# Insight:
# The correlation heatmap highlights the strength and direction of relationships between numerical variables in the dataset.
```
